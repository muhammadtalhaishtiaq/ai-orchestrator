## What are we learning?

Univariate analysis is the simplest form of data exploration: we examine one variable at a time to understand its distribution, central tendency, and spread. It’s the first step before asking how variables relate to each other.

## The idea in plain English

Imagine you receive a big bag of apples and want to know how heavy they are. You weigh each apple and then look at the list of weights alone—no colors, no prices, just weights. Univariate analysis is exactly that: studying one column of data (the weights) to find its typical value, range, and shape.

## Setup

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_style('whitegrid')
print('Setup done!')
```

## Step 1 — Load data

```python
df = sns.load_dataset('mpg').dropna()
print(df.head())
print('Shape:', df.shape)
```

## Step 2 — Apply Univariate Analysis

```python
# Focus on one variable: miles per gallon (mpg)
mpg = df['mpg']
print('Five-number summary:')
print(mpg.describe())
print('\nMode:', mpg.mode().values)
print('Variance:', mpg.var().round(2))
print('Skewness:', mpg.skew().round(2))
```

## Step 3 — Visualise

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Histogram with KDE
sns.histplot(mpg, bins=15, kde=True, ax=axes[0], color='royalblue')
axes[0].set_title('Distribution of MPG')
axes[0].set_xlabel('Miles per Gallon')

# Boxplot
sns.boxplot(x=mpg, ax=axes[1], color='coral')
axes[1].set_title('Boxplot of MPG')
axes[1].set_xlabel('Miles per Gallon')

plt.tight_layout()
plt.show()
```

## Results & interpretation

```python
# Key take-aways
mean_mpg = mpg.mean()
median_mpg = mpg.median()
iqr = mpg.quantile(0.75) - mpg.quantile(0.25)

print(f'Mean MPG: {mean_mpg:.1f}')
print(f'Median MPG: {median_mpg:.1f}')
print(f'IQR: {iqr:.1f}')
print(f'Shape: {"Right-skewed" if mpg.skew() > 0.5 else "Roughly symmetric"}')
```

## Summary

- Univariate analysis inspects a single variable in isolation.
- We computed descriptive statistics (mean, median, variance, skewness).
- Histograms and boxplots quickly reveal distribution shape and outliers.
- The `mpg` variable is slightly right-skewed with most cars between 15–30 MPG.
- These insights set the stage for deeper multivariate questions.
