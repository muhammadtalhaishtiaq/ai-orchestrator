## 1. Derivatives Basics

```python
# Derivative = rate of change at a point
# For f(x) = x², derivative f'(x) = 2x

def f(x):
    return x**2

# Numerical derivative approximation
def numerical_derivative(func, x, h=1e-5):
    return (func(x + h) - func(x - h)) / (2 * h)

# Analytical derivative
def f_derivative(x):
    return 2 * x

# Test
x_test = 3.0
num_deriv = numerical_derivative(f, x_test)
ana_deriv = f_derivative(x_test)

print(f"At x = {x_test}:")
print(f"Numerical derivative: {num_deriv:.6f}")
print(f"Analytical derivative: {ana_deriv:.6f}")
print(f"Match? {np.isclose(num_deriv, ana_deriv)}")
```

```python
# Visualize: function and its derivative
x = np.linspace(-4, 4, 100)
y = f(x)
y_prime = f_derivative(x)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

# Function
ax1.plot(x, y, linewidth=2, label='f(x) = x²')
ax1.scatter([x_test], [f(x_test)], color='red', s=100, zorder=5)
ax1.set_title('Function f(x) = x²')
ax1.set_xlabel('x')
ax1.set_ylabel('f(x)')
ax1.grid(True, alpha=0.3)
ax1.legend()

# Derivative
ax2.plot(x, y_prime, linewidth=2, color='orange', label="f'(x) = 2x")
ax2.scatter([x_test], [f_derivative(x_test)], color='red', s=100, zorder=5)
ax2.axhline(y=0, color='k', linestyle='-', alpha=0.3)
ax2.set_title('Derivative (Rate of Change)')
ax2.set_xlabel('x')
ax2.set_ylabel("f'(x)")
ax2.grid(True, alpha=0.3)
ax2.legend()

plt.tight_layout()
plt.show()

print(f"At x={x_test}, slope = {f_derivative(x_test)} (steep climb!)")
```

## 2. Chain Rule (THE MOST IMPORTANT!)

```python
# Chain rule: df/dx = df/du * du/dx
# For f(g(x)): f'(g(x)) = f'(u) * g'(x) where u = g(x)

# Example: f(x) = (2x + 3)²
# Let u = 2x + 3, then f = u²
# df/dx = df/du * du/dx = (2u) * 2 = 4(2x + 3)

def f_composite(x):
    return (2*x + 3)**2

def f_composite_derivative(x):
    # Using chain rule
    return 4 * (2*x + 3)

x_test = 2.0
num_deriv = numerical_derivative(f_composite, x_test)
ana_deriv = f_composite_derivative(x_test)

print(f"Function: f(x) = (2x + 3)²")
print(f"Using chain rule: f'(x) = 4(2x + 3)")
print(f"\nAt x = {x_test}:")
print(f"Numerical: {num_deriv:.6f}")
print(f"Analytical: {ana_deriv:.6f}")
```

```python
# Chain rule visualization: how changes propagate
print("\n=== Chain Rule Breakdown ===")
print(f"At x = {x_test}:")
print(f"  u = 2x + 3 = 2({x_test}) + 3 = {2*x_test + 3}")
print(f"  f(u) = u² = {(2*x_test + 3)**2}")
print(f"  du/dx = 2")
print(f"  df/du = 2u = 2({2*x_test + 3}) = {2*(2*x_test + 3)}")
print(f"  df/dx = df/du × du/dx = {2*(2*x_test + 3)} × 2 = {f_composite_derivative(x_test)}")
```

## 3. Partial Derivatives

```python
# Partial derivative: derivative w.r.t. one variable, holding others constant
# f(x, y) = x² + 3xy + y²
# ∂f/∂x = 2x + 3y  (treat y as constant)
# ∂f/∂y = 3x + 2y  (treat x as constant)

def f_2d(x, y):
    return x**2 + 3*x*y + y**2

def df_dx(x, y):
    return 2*x + 3*y

def df_dy(x, y):
    return 3*x + 2*y

x_test, y_test = 1.0, 2.0

print(f"f(x, y) = x² + 3xy + y²")
print(f"\nAt (x, y) = ({x_test}, {y_test}):")
print(f"f = {f_2d(x_test, y_test)}")
print(f"∂f/∂x = {df_dx(x_test, y_test)}")
print(f"∂f/∂y = {df_dy(x_test, y_test)}")
```

```python
# Visualize 2D function and gradients
x_range = np.linspace(-3, 3, 50)
y_range = np.linspace(-3, 3, 50)
X, Y = np.meshgrid(x_range, y_range)
Z = f_2d(X, Y)

fig = plt.figure(figsize=(12, 5))

# Contour plot with gradient vectors
ax1 = fig.add_subplot(121)
contour = ax1.contour(X, Y, Z, levels=15)
ax1.clabel(contour, inline=True, fontsize=8)

# Add gradient vectors at several points
x_sample = np.array([1, 1, 2, -1])
y_sample = np.array([1, -1, 1, 1])
dx = df_dx(x_sample, y_sample)
dy = df_dy(x_sample, y_sample)
ax1.quiver(x_sample, y_sample, dx, dy, scale=100)

ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.set_title('Function Contours with Gradients')
ax1.grid(True, alpha=0.3)

# 3D surface
ax2 = fig.add_subplot(122, projection='3d')
ax2.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8)
ax2.scatter([x_test], [y_test], [f_2d(x_test, y_test)], color='red', s=100)
ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.set_zlabel('f(x, y)')
ax2.set_title('3D Surface')

plt.tight_layout()
plt.show()
```

## 4. Gradient Vector

```python
# Gradient = vector of all partial derivatives
# ∇f = [∂f/∂x, ∂f/∂y, ...]
# Points in direction of STEEPEST ASCENT

def gradient_2d(x, y):
    return np.array([df_dx(x, y), df_dy(x, y)])

x, y = 1.0, 2.0
grad = gradient_2d(x, y)

print(f"At point ({x}, {y}):")
print(f"Gradient ∇f = {grad}")
print(f"Direction of steepest increase")
print(f"Magnitude: {np.linalg.norm(grad):.4f}")
print(f"Unit gradient: {grad / np.linalg.norm(grad)}")
```

## 5. Gradient Descent (How Neural Networks Learn!)

```python
# Minimize f(x) = (x - 5)² using gradient descent
# Start at random point, move opposite to gradient (downhill)

def loss_func(x):
    return (x - 5)**2

def loss_gradient(x):
    return 2 * (x - 5)

# Gradient descent
x = 0.0  # Start point
learning_rate = 0.1
iterations = 20
history = [x]
loss_history = [loss_func(x)]

print("Gradient Descent Progress:")
print(f"{'Iter':<5} {'x':<10} {'Loss':<10} {'Gradient':<10}")
print("-" * 35)

for i in range(iterations):
    grad = loss_gradient(x)
    x = x - learning_rate * grad  # Move opposite to gradient
    loss = loss_func(x)
    history.append(x)
    loss_history.append(loss)
    
    if i % 4 == 0 or i == iterations - 1:
        print(f"{i:<5} {x:<10.4f} {loss:<10.6f} {grad:<10.4f}")

print(f"\nFinal x: {x:.6f} (should be ≈ 5)")
```

```python
# Visualize gradient descent journey
x_range = np.linspace(-2, 8, 100)
y_range = loss_func(x_range)

plt.figure(figsize=(10, 6))
plt.plot(x_range, y_range, 'b-', linewidth=2, label='Loss Function')
plt.plot(history, loss_history, 'r.-', linewidth=2, markersize=8, label='GD Path')
plt.scatter([5], [0], color='green', s=200, marker='*', label='Minimum')

plt.xlabel('x')
plt.ylabel('Loss')
plt.title('Gradient Descent Journey')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print("\nThis is HOW neural networks learn - by computing gradients and moving downhill!")
```

## Key Takeaways

1. **Derivative** = Rate of change at a point
2. **Chain Rule** = How to differentiate composite functions (neural networks!)
3. **Gradient** = Vector pointing to steepest increase
4. **Gradient Descent** = Algorithm that follows negative gradient to minimize loss
5. **Backpropagation** = Chain rule applied backward through network layers

**All of deep learning is built on these concepts!**
