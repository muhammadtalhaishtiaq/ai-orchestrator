## 1. Lists and Dictionaries

```python
# Lists - your bread and butter
numbers = [1, 2, 3, 4, 5]
print(f"First: {numbers[0]}, Last: {numbers[-1]}")
print(f"Slice [1:3]: {numbers[1:3]}")

# Common operations
numbers.append(6)
numbers.extend([7, 8])
print(f"After append/extend: {numbers}")
```

```python
# Dictionaries - key-value pairs everywhere in ML
model_config = {
    'learning_rate': 0.01,
    'epochs': 100,
    'batch_size': 32
}

print(f"Learning rate: {model_config['learning_rate']}")
print(f"All keys: {list(model_config.keys())}")

# Safe access with .get()
dropout = model_config.get('dropout', 0.5)  # Returns 0.5 if key missing
print(f"Dropout (default): {dropout}")
```

## 2. List Comprehensions

Write less, do more. You'll see these everywhere in ML code.

```python
# Instead of this...
squared = []
for x in range(10):
    squared.append(x ** 2)
print(f"Loop way: {squared}")

# Do this
squared = [x ** 2 for x in range(10)]
print(f"Comprehension: {squared}")

# With conditions
evens = [x for x in range(20) if x % 2 == 0]
print(f"Even numbers: {evens}")
```

```python
# Dictionary comprehension - super useful for feature mapping
words = ['cat', 'dog', 'bird']
word_to_id = {word: idx for idx, word in enumerate(words)}
print(f"Word mapping: {word_to_id}")

# Reverse it
id_to_word = {idx: word for word, idx in word_to_id.items()}
print(f"ID mapping: {id_to_word}")
```

## 3. Functions

```python
# Basic function with type hints (good practice)
def calculate_mse(y_true: list, y_pred: list) -> float:
    """Calculate Mean Squared Error."""
    n = len(y_true)
    return sum((t - p) ** 2 for t, p in zip(y_true, y_pred)) / n

# Test it
actual = [3, 5, 2, 7]
predicted = [2.5, 5.5, 2, 8]
print(f"MSE: {calculate_mse(actual, predicted):.4f}")
```

```python
# Default arguments and *args, **kwargs
def train_model(X, y, epochs=10, learning_rate=0.01, **kwargs):
    """Flexible function signature - common in ML libraries."""
    print(f"Training for {epochs} epochs at lr={learning_rate}")
    print(f"Extra params: {kwargs}")

train_model([1,2,3], [0,1,0], epochs=50, verbose=True, early_stopping=True)
```

```python
# Lambda functions - quick one-liners
square = lambda x: x ** 2
print(f"Square of 5: {square(5)}")

# Often used with map, filter, sorted
data = [{'name': 'Alice', 'score': 85}, {'name': 'Bob', 'score': 92}]
sorted_data = sorted(data, key=lambda x: x['score'], reverse=True)
print(f"Sorted by score: {sorted_data}")
```

## 4. Classes (Just Enough)

You'll create custom models/transformers. Here's the minimum you need.

```python
class SimpleScaler:
    """A basic min-max scaler - like sklearn's MinMaxScaler."""
    
    def __init__(self):
        self.min_ = None
        self.max_ = None
    
    def fit(self, X):
        """Learn min and max from data."""
        self.min_ = min(X)
        self.max_ = max(X)
        return self
    
    def transform(self, X):
        """Scale data to 0-1 range."""
        return [(x - self.min_) / (self.max_ - self.min_) for x in X]
    
    def fit_transform(self, X):
        """Fit and transform in one step."""
        return self.fit(X).transform(X)

# Use it
scaler = SimpleScaler()
data = [10, 20, 30, 40, 50]
scaled = scaler.fit_transform(data)
print(f"Original: {data}")
print(f"Scaled: {scaled}")
```

## 5. File Handling

```python
import json

# Save model config
config = {'model': 'RandomForest', 'n_estimators': 100, 'max_depth': 10}

with open('config.json', 'w') as f:
    json.dump(config, f, indent=2)
print("Saved config.json")

# Load it back
with open('config.json', 'r') as f:
    loaded = json.load(f)
print(f"Loaded: {loaded}")

# Cleanup
import os
os.remove('config.json')
```
