## What are we learning?

Machine-learning models only understand numbers.  In this notebook you’ll learn how to turn text categories (like "red", "green", "blue") into numbers that algorithms can digest—without accidentally inventing an order that isn’t there.

## The idea in plain English

Imagine a fruit basket with apples, bananas and cherries.  If we label them 1, 2, 3 the model might think cherries are “bigger” than apples.  One-hot encoding is like giving each fruit its own light-switch: apple [1 0 0], banana [0 1 0], cherry [0 0 1].  No fruit is “larger”; we just indicate presence.

## Setup

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
import matplotlib.pyplot as plt
import seaborn as sns

print('Setup done!')
```

## Step 1 — Load data

```python
df = pd.read_csv('https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv')
df.head()
```

## Step 2 — Apply Encoding Categorical

```python
# Separate numeric and categorical columns
num_cols = ['total_bill', 'tip', 'size']
cat_cols = ['sex', 'smoker', 'day', 'time']

X_cat = df[cat_cols]
print("Original categorical data:")
X_cat.head()
```

```python
# One-hot encode the categorical features
ohe = OneHotEncoder(sparse=False, drop='first')  # drop first to avoid dummy trap
X_ohe = ohe.fit_transform(X_cat)
feature_names = ohe.get_feature_names_out(cat_cols)
X_encoded = pd.DataFrame(X_ohe, columns=feature_names)
X_encoded.head()
```

## Step 3 — Visualise

```python
# Heat-map of the encoded matrix
plt.figure(figsize=(8, 4))
sns.heatmap(X_encoded.iloc[:20], annot=True, fmt='.0f', cmap='Blues')
plt.title('One-Hot Encoded Categories (first 20 rows)')
plt.show()
```

## Results & interpretation

```python
# Check shape change
print(f"Original categorical shape: {X_cat.shape}")
print(f"After one-hot encoding:      {X_encoded.shape}")
print("\nNew numeric columns created:")
print(list(X_encoded.columns))
```

## Summary

- Categorical text must be converted to numbers before modelling.
- One-hot encoding creates binary (0/1) columns for each category level.
- Dropping one column per feature avoids the dummy-variable trap.
- The resulting numeric matrix can be fed directly into scikit-learn models.
