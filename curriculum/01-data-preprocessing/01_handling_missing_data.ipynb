## What are we learning?

Missing values are like holes in a puzzle—they break algorithms and distort insights. In this notebook you’ll learn how to detect, measure, and repair them with pandas and scikit-learn so your models train on complete, reliable data.

## The idea in plain English

Imagine a spreadsheet where some cells are blank. If you average a column with blanks, the result is wrong. Instead of deleting the whole row (throwing away good data), we can *fill* the blanks with educated guesses—column averages, medians, or even predicted values—so every row stays usable.

## Setup

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.impute import SimpleImputer, KNNImputer

sns.set_theme(style="whitegrid")
print('Setup done!')
```

## Step 1 — Load data

```python
# Toy dataset: 200 customers with Age, Income, and SpendScore
df = pd.read_csv('https://raw.githubusercontent.com/selva86/datasets/master/MallCustomers_missing.csv')
print('Shape:', df.shape)
df.head()
```

## Step 2 — Apply Handling Missing Data

```python
# 1. Detect missing values
print('Missing counts:\n', df.isna().sum())
print('Missing %:\n', (df.isna().mean()*100).round(1))
```

```python
# 2. Strategy 1: Drop rows with any missing values
df_drop = df.dropna()
print('Rows after dropna:', df_drop.shape[0])
```

```python
# 3. Strategy 2: Mean imputation for numeric columns
mean_imp = SimpleImputer(strategy='mean')
df_mean = df.copy()
df_mean[['Age','Annual Income (k$)','Spending Score (1-100)']] = mean_imp.fit_transform(df_mean[['Age','Annual Income (k$)','Spending Score (1-100)']])
print('After mean imputation:')
print(df_mean.isna().sum())
```

```python
# 4. Strategy 3: KNN imputation (k=5) uses similar rows
knn_imp = KNNImputer(n_neighbors=5)
df_knn = df.copy()
df_knn[['Age','Annual Income (k$)','Spending Score (1-100)']] = knn_imp.fit_transform(df_knn[['Age','Annual Income (k$)','Spending Score (1-100)']])
print('After KNN imputation:')
print(df_knn.isna().sum())
```

## Step 3 — Visualise

```python
# Compare Age distributions after each strategy
plt.figure(figsize=(9,5))
sns.kdeplot(df['Age'], label='Original (with NaN)', color='red', ls='--')
sns.kdeplot(df_mean['Age'], label='Mean imputed', color='blue')
sns.kdeplot(df_knn['Age'], label='KNN imputed', color='green')
plt.title('Age Distribution: Effect of Imputation')
plt.xlabel('Age')
plt.ylabel('Density')
plt.legend()
plt.show()
```

## Results & interpretation

```python
# How much data we keep
print('Original rows:', len(df))
print('Rows kept by dropna:', len(df_drop))
print('Rows kept by imputation:', len(df_mean))
print('Data retention: {:.1%}'.format(len(df_mean)/len(df)))
```

```python
# Compare means to see bias introduced
print('Original column means (NaN excluded):')
print(df.mean().round(2))
print('\nMean-imputed column means:')
print(df_mean.mean().round(2))
print('\nKNN-imputed column means:')
print(df_knn.mean().round(2))
```

## Summary

- `isna()` and `isna().mean()` quickly reveal how much data is missing.
- Dropping rows is simple but can discard useful information.
- Mean imputation preserves row count but may reduce variance; KNN imputation preserves relationships between variables.
- Always visualise distributions before and after imputation to check for distortion.
