## ⚙️ Setup

Just three libraries — nothing fancy.

```python
import numpy as np          # numbers and arrays
import pandas as pd         # tables (DataFrames)
import matplotlib.pyplot as plt  # charts
import seaborn as sns        # nicer charts

plt.style.use('seaborn-v0_8-whitegrid')
print('✅ Setup done!')
```

## 📦 Step 1 — Load the data

We'll use a small, realistic dataset so you can see exactly what's happening. Think of this as a hiring dataset — age, salary, experience, hired or not.

```python
# Sample dataset — realistic but small enough to understand completely
data = {
    'age':        [25, 30, 35, None, 45, 50, 28, None, 40, 33],
    'salary':     [50000, 60000, None, 80000, 90000, 100000, 55000, 75000, None, 65000],
    'experience': [2, 5, 8, 12, 15, 20, 3, 10, 14, 7],
    'hired':      [0, 1, 1, 1, 1, 1, 0, 1, 1, 1]
}

df = pd.DataFrame(data)
print(f'Dataset shape: {df.shape}')
print('\nFirst 5 rows:')
df.head()
```

## 🔨 Step 2 — Apply Feature Extraction

Let's look at what we're working with before making changes.

```python
# Always inspect first — never change data blindly
print('Missing values per column:')
print(df.isnull().sum())

print('\nData types:')
print(df.dtypes)

print('\nBasic stats:')
df.describe()
```

### Step 3 — Fix it

```python
# Fill missing numerical values with the median
# (median is better than mean when you have outliers)
df_clean = df.fillna(df.median(numeric_only=True))

print('After cleaning:')
print('Missing values:', df_clean.isnull().sum().sum())
df_clean.head()
```

## 📊 What did we get?

Let's check the before vs after.

```python
print('BEFORE — missing values:')
print(df.isnull().sum())

print('\nAFTER — missing values:')
print(df_clean.isnull().sum())

print('\n✅ Data is clean and ready for a model!')
```

## 📈 Let's see it

Visuals make it click faster than words.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Left chart: where data was missing
missing_counts = df.isnull().sum()
axes[0].bar(missing_counts.index, missing_counts.values, color='#ef4444', alpha=0.8)
axes[0].set_title('Missing Values — Before Cleaning', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Column')
axes[0].set_ylabel('Missing Count')

# Right chart: salary distribution after cleaning
axes[1].hist(df_clean['salary'], bins=8, color='#22c55e', edgecolor='white', alpha=0.9)
axes[1].set_title('Salary Distribution — After Cleaning', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Salary ($)')
axes[1].set_ylabel('Count')

plt.tight_layout()
plt.show()
```
