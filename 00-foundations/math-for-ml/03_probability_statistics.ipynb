## Why Probability?

ML is about making predictions with uncertainty. Understanding probability means understanding why models give confidence scores, how to handle uncertainty, and Bayesian methods.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

np.random.seed(42)
print("Ready to explore probability!")
```

---
## 1. Probability Basics

```python
# Basic probability
# P(A) = probability of event A
# Range: [0, 1]

# Example: Fair die roll
total_outcomes = 6
favorable = 1  # Rolling a 3

p_three = favorable / total_outcomes
print(f"P(rolling a 3) = {p_three:.3f}")

# Multiple outcomes
p_even = 3 / 6  # 2, 4, 6
print(f"P(rolling even) = {p_even:.3f}")

# Complement
p_not_even = 1 - p_even
print(f"P(rolling odd) = {p_not_even:.3f}")
```

```python
# Joint probability: P(A AND B)
# For independent events: P(A ∩ B) = P(A) × P(B)

# Example: Coin + Die
p_heads = 0.5
p_six = 1/6

p_heads_and_six = p_heads * p_six
print(f"P(heads AND six) = {p_heads_and_six:.4f}")
print(f"or 1 in {int(1/p_heads_and_six)} outcomes")
```

---
## 2. Conditional Probability

```python
# Conditional: P(A|B) = "probability of A given B happened"
# P(A|B) = P(A ∩ B) / P(B)

# Example: Disease testing
# Disease prevalence in population: 1%
# Test accuracy: 99%

p_disease = 0.01
p_test_positive_given_disease = 0.99  # Sensitivity
p_test_positive_given_no_disease = 0.01  # False positive rate

# Someone tests positive. What's probability they have disease?
# P(disease | positive) = P(positive | disease) × P(disease) / P(positive)

p_has_disease = p_disease
p_no_disease = 1 - p_disease

p_positive = (p_test_positive_given_disease * p_has_disease + 
              p_test_positive_given_no_disease * p_no_disease)

p_disease_given_positive = (p_test_positive_given_disease * p_has_disease) / p_positive

print(f"Prior (base rate): P(disease) = {p_disease:.3f}")
print(f"Posterior: P(disease | positive test) = {p_disease_given_positive:.3f}")
print(f"\nSurprising! Test is 99% accurate but...")
print(f"If you test positive, only {p_disease_given_positive*100:.1f}% likely you have it!")
print(f"(Because disease is rare)")
```

---
## 3. Bayes' Theorem (The Foundation!)

```python
# Bayes' Theorem:
# P(A|B) = P(B|A) × P(A) / P(B)

# Components:
# P(A|B) = Posterior (what we want to know)
# P(B|A) = Likelihood (evidence given hypothesis)
# P(A) = Prior (initial belief)
# P(B) = Marginal likelihood (normalization)

print("Bayes' Theorem: P(A|B) = P(B|A) × P(A) / P(B)")
print()
print("Components:")
print(f"  P(A|B) = Posterior (disease | positive test) = {p_disease_given_positive:.3f}")
print(f"  P(B|A) = Likelihood (positive | disease) = {p_test_positive_given_disease:.3f}")
print(f"  P(A) = Prior (disease prevalence) = {p_has_disease:.3f}")
print(f"  P(B) = Marginal (any positive test) = {p_positive:.3f}")
```

---
## 4. Expected Value & Variance

```python
# Expected value E[X] = mean value (long run average)
# E[X] = Σ(x × P(x))

# Example: Dice roll
outcomes = np.array([1, 2, 3, 4, 5, 6])
probabilities = np.ones(6) / 6  # Fair die

expected_value = np.sum(outcomes * probabilities)
print(f"Fair die expected value: {expected_value:.3f}")
print(f"Manual: (1+2+3+4+5+6)/6 = {sum(outcomes)/6}")
```

```python
# Variance = spread = how far from expected
# Var(X) = E[(X - E[X])²]

mean = expected_value
variance = np.sum((outcomes - mean)**2 * probabilities)
std_dev = np.sqrt(variance)

print(f"Mean (E[X]): {mean:.3f}")
print(f"Variance: {variance:.3f}")
print(f"Std Dev: {std_dev:.3f}")
print(f"\nInterpretation: Most rolls within ±{std_dev:.1f} of mean {mean:.1f}")
```

## 5. Common Distributions

```python
# Bernoulli: single binary outcome (success/failure)
p = 0.3
bernoulli_samples = np.random.binomial(n=1, p=p, size=1000)

print(f"Bernoulli(p={p}):")
print(f"  Mean (theoretical): {p}")
print(f"  Mean (observed): {bernoulli_samples.mean():.3f}")
```

```python
# Binomial: number of successes in n trials
n_trials = 10
p_success = 0.3
binomial_samples = np.random.binomial(n=n_trials, p=p_success, size=10000)

print(f"Binomial(n={n_trials}, p={p_success}):")
print(f"  Mean: {binomial_samples.mean():.2f}")
print(f"  Theoretical: {n_trials * p_success}")
```

```python
# Normal (Gaussian): bell curve
mu = 100
sigma = 15
normal_samples = np.random.normal(mu, sigma, 100000)

print(f"Normal(μ={mu}, σ={sigma}):")
print(f"  Sample mean: {normal_samples.mean():.2f}")
print(f"  Sample std: {normal_samples.std():.2f}")
```

```python
# Visualize distributions
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Bernoulli
unique, counts = np.unique(bernoulli_samples, return_counts=True)
axes[0, 0].bar(unique, counts/len(bernoulli_samples))
axes[0, 0].set_title('Bernoulli(p=0.3)')
axes[0, 0].set_ylabel('Probability')
axes[0, 0].set_xticks([0, 1])

# Binomial
axes[0, 1].hist(binomial_samples, bins=12, density=True, alpha=0.7, edgecolor='black')
axes[0, 1].set_title(f'Binomial(n={n_trials}, p={p_success})')
axes[0, 1].set_ylabel('Probability')

# Normal
axes[1, 0].hist(normal_samples, bins=50, density=True, alpha=0.7, edgecolor='black')
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 100)
axes[1, 0].plot(x, stats.norm.pdf(x, mu, sigma), 'r-', linewidth=2)
axes[1, 0].set_title(f'Normal(μ={mu}, σ={sigma})')
axes[1, 0].set_ylabel('Probability Density')

# Q-Q plot (Normal check)
stats.probplot(normal_samples, dist="norm", plot=axes[1, 1])
axes[1, 1].set_title('Q-Q Plot (Normal Distribution Check)')

plt.tight_layout()
plt.show()
```

## 6. Central Limit Theorem (Why Normal Distribution is Everywhere)

```python
# Central Limit Theorem: Average of independent samples → Normal distribution
# Even if original distribution is weird!

# Start with uniform (not normal)
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

# Original distribution
uniform_data = np.random.uniform(0, 10, 100000)
axes[0].hist(uniform_data, bins=50, edgecolor='black', alpha=0.7)
axes[0].set_title('Original: Uniform Distribution')
axes[0].set_ylabel('Frequency')

# Means of samples (n=10)
means_10 = [np.mean(np.random.uniform(0, 10, 10)) for _ in range(10000)]
axes[1].hist(means_10, bins=50, edgecolor='black', alpha=0.7)
axes[1].set_title('Average of 10 Samples')
axes[1].set_ylabel('Frequency')

# Means of samples (n=100)
means_100 = [np.mean(np.random.uniform(0, 10, 100)) for _ in range(10000)]
axes[2].hist(means_100, bins=50, edgecolor='black', alpha=0.7)
axes[2].set_title('Average of 100 Samples')
axes[2].set_ylabel('Frequency')

plt.suptitle('Central Limit Theorem: Distributions of Means → Normal', fontsize=14)
plt.tight_layout()
plt.show()

print("Even though original is uniform, averages form bell curve!")
```
