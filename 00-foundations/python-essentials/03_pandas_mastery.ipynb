## Why Pandas?

Real data comes in tables (CSV, Excel, databases). Pandas is how you load it, clean it, and prep it for ML. You'll spend 80% of your time here.

```python
import pandas as pd
import numpy as np

print(f"Pandas version: {pd.__version__}")
```

---
## 1. Creating DataFrames

```python
# From dictionary - most common way
data = {
    'name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'age': [25, 30, 35, 28],
    'salary': [50000, 60000, 75000, 55000],
    'department': ['Engineering', 'Marketing', 'Engineering', 'Sales']
}

df = pd.DataFrame(data)
```

```python
# Quick look at the data
print(f"Shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
print(f"\nData types:\n{df.dtypes}")
```

```python
# Load from CSV (most common in real projects)
# df = pd.read_csv('data.csv')

# We'll create a sample dataset
np.random.seed(42)
n = 100

df = pd.DataFrame({
    'age': np.random.randint(20, 60, n),
    'income': np.random.randint(30000, 120000, n),
    'experience': np.random.randint(0, 30, n),
    'education': np.random.choice(['High School', 'Bachelor', 'Master', 'PhD'], n),
    'purchased': np.random.choice([0, 1], n)
})

df.head(10)
```

---
## 2. Exploring Data

```python
# First and last rows
print("First 3 rows:")
display(df.head(3))

print("\nLast 3 rows:")
display(df.tail(3))
```

```python
# Summary stats - your first step with any dataset
df.describe()
```

```python
# Info about columns and missing values
df.info()
```

```python
# Value counts for categorical columns
df['education'].value_counts()
```

---
## 3. Selecting Data

```python
# Single column (returns Series)
ages = df['age']
print(f"Type: {type(ages)}")
print(f"First 5: {ages.head().tolist()}")
```

```python
# Multiple columns (returns DataFrame)
subset = df[['age', 'income', 'purchased']]
subset.head()
```

```python
# loc - select by label
# iloc - select by position

# First 5 rows, specific columns
df.loc[:4, ['age', 'income']]
```

```python
# By position
df.iloc[0:5, 0:3]  # First 5 rows, first 3 columns
```

---
## 4. Filtering Data

```python
# Simple filter
high_income = df[df['income'] > 80000]
print(f"High income count: {len(high_income)}")
high_income.head()
```

```python
# Multiple conditions (use & for AND, | for OR)
filtered = df[(df['age'] > 30) & (df['income'] > 60000)]
print(f"Age > 30 AND income > 60k: {len(filtered)} rows")
filtered.head()
```

```python
# Filter with isin()
advanced = df[df['education'].isin(['Master', 'PhD'])]
print(f"Advanced degrees: {len(advanced)} rows")
```

---
## 5. Creating New Columns

```python
# Simple calculation
df['income_per_year_exp'] = df['income'] / (df['experience'] + 1)

# Conditional column
df['income_level'] = df['income'].apply(
    lambda x: 'High' if x > 80000 else 'Medium' if x > 50000 else 'Low'
)

df[['income', 'experience', 'income_per_year_exp', 'income_level']].head(10)
```

```python
# Using np.where (like if-else)
df['is_senior'] = np.where(df['age'] >= 40, 1, 0)
df[['age', 'is_senior']].head(10)
```

---
## 6. GroupBy - Aggregate by Category

```python
# Average income by education level
df.groupby('education')['income'].mean().sort_values(ascending=False)
```

```python
# Multiple aggregations
df.groupby('education').agg({
    'income': ['mean', 'std', 'count'],
    'age': 'mean'
}).round(2)
```

```python
# Crosstab - count combinations
pd.crosstab(df['education'], df['purchased'])
```

---
## 7. Handling Missing Values

```python
# Add some missing values for demo
df_missing = df.copy()
df_missing.loc[0:5, 'income'] = np.nan
df_missing.loc[10:15, 'age'] = np.nan

# Check for missing values
print("Missing values per column:")
print(df_missing.isnull().sum())
```

```python
# Fill with mean
df_filled = df_missing.copy()
df_filled['income'] = df_filled['income'].fillna(df_filled['income'].mean())
df_filled['age'] = df_filled['age'].fillna(df_filled['age'].median())

print("After filling:")
print(df_filled.isnull().sum())
```
